# 🧬 Part 2: Multi-Scale CNN Classifier Training

This notebook serves as **Part 2** of our deep learning workflow. It loads the pre-extracted DNABERT-2 embeddings from disk, splits them into stratified train/val sets, and trains the Multi-Scale CNN (mCNN) classifier.

## Advantages of This Separation
1. **Super Fast Training**: Since we do not run the large DNABERT-2 model during training epochs, each epoch takes less than **1 second** to complete.
2. **Zero OOM Risk**: The extraction notebook holds the weights, whereas this notebook only processes static NumPy arrays using memory-sharing loaders, which keeps system RAM consumption extremely low.
3. **Reproducibility**: We can tune hyperparameters of the mCNN (learning rate, layers, dropout) without re-extracting representations.

### 1. Installation and Repository Setup
